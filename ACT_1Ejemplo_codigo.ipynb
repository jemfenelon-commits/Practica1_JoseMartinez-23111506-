En esta actividad usarás el archivo de datos `Forbes_2000_top_company.csv`. Este conjunto contiene las 2000 principales empresas, clasificadas mediante el uso de cuatro métricas: ventas, ganancias, activos y valor de mercado. Tiene las 11 columnas de datos:

*   `2022 Ranking`: Ranking del año actual de la empresa
*   `Organization Name`: Nombre
*   `Industry`: El tipo de industria con la que trata principalmente
*   `Country`: País de origen
*   `Year Founded`: Año en que se fundó
*   `CEO`: CEO
*   `Profits (Billions)`: Ganancias del año
*   `Assets (Billions)`: Activos del año
*   `Market Value (Billions)`: Valor de mercado del año
*   `Total Employees`: Total de empleados

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar el estilo visual
sns.set_theme(style="whitegrid")

# =====================================================================
# 1. MAPA DE CALOR (1 Gráfica)
# =====================================================================
plt.figure(figsize=(10, 8))
numeric_cols = forbes_df.select_dtypes(include=[np.number])
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Mapa de Calor: Correlación de Variables Numéricas')
plt.show()

# =====================================================================
# 2. DIAGRAMAS DE DISPERSIÓN (3 Gráficas)
# =====================================================================
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Ganancias vs Valor de Mercado
sns.scatterplot(data=forbes_df, x='Profits (Billions)', y='Market Value (Billions)', ax=axes[0], color='blue', alpha=0.6)
axes[0].set_title('Ganancias vs Valor de Mercado')

# Ingresos vs Activos
sns.scatterplot(data=forbes_df, x='Revenue (Billions)', y='Assets (Billions)', ax=axes[1], color='green', alpha=0.6)
axes[1].set_title('Ingresos vs Activos')

# Total de Empleados vs Ingresos
sns.scatterplot(data=forbes_df, x='Total Employees', y='Revenue (Billions)', ax=axes[2], color='red', alpha=0.6)
axes[2].set_title('Empleados vs Ingresos')

plt.tight_layout()
plt.show()

# =====================================================================
# 3. DIAGRAMAS DE CAJA / BOXPLOTS (2 Gráficas)
# =====================================================================
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(data=forbes_df, y='Profits (Billions)', ax=axes[0], color='cyan')
axes[0].set_title('Distribución de Ganancias (Billones)')
axes[0].set_ylim(-5, forbes_df['Profits (Billions)'].quantile(0.90))

sns.boxplot(data=forbes_df, y='Revenue (Billions)', ax=axes[1], color='magenta')
axes[1].set_title('Distribución de Ingresos (Billones)')
axes[1].set_ylim(-5, forbes_df['Revenue (Billions)'].quantile(0.90))

plt.tight_layout()
plt.show()

# =====================================================================
# 4. GRÁFICAS DE BARRAS DATOS CATEGÓRICOS (5 Gráficas)
# =====================================================================
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

# Gráfica 1: Top 10 Países con más empresas
forbes_df['Country'].value_counts().head(10).plot(kind='bar', ax=axes[0], color='skyblue')
axes[0].set_title('Top 10 Países')
axes[0].set_ylabel('Cantidad de Empresas')
axes[0].tick_params(axis='x', rotation=45)

# Gráfica 2: Top 10 Industrias predominantes
forbes_df['Industry'].value_counts().head(10).plot(kind='bar', ax=axes[1], color='lightgreen')
axes[1].set_title('Top 10 Industrias')
axes[1].tick_params(axis='x', rotation=45)

# Gráfica 3: Top 10 Años de fundación
if 'Year Founded' in forbes_df.columns:
    forbes_df[forbes_df['Year Founded'] > 0]['Year Founded'].value_counts().head(10).plot(kind='bar', ax=axes[2], color='orange')
    axes[2].set_title('Top 10 Años de Fundación')
    axes[2].tick_params(axis='x', rotation=45)

# Gráfica 4: Top 5 Empresas con mayores ganancias
top_profits = forbes_df.nlargest(5, 'Profits (Billions)')
sns.barplot(data=top_profits, x='Organization Name', y='Profits (Billions)', ax=axes[3], palette='Blues_r')
axes[3].set_title('Top 5 Empresas por Ganancias')
axes[3].tick_params(axis='x', rotation=45)

# Gráfica 5: Top 5 Empresas con mayor número de empleados
top_emp = forbes_df.nlargest(5, 'Total Employees')
sns.barplot(data=top_emp, x='Organization Name', y='Total Employees', ax=axes[4], palette='Greens_r')
axes[4].set_title('Top 5 Empresas por Número de Empleados')
axes[4].tick_params(axis='x', rotation=45)

# Ocultar el sexto panel vacío
axes[5].axis('off')

plt.tight_layout()
plt.show()
```
